```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
cars = pd.read_csv("car-sales.csv")
cars
```

```python
make_counts = cars["Make"].value_counts()

# Plot pie chart
plt.figure(figsize=(6, 6))
plt.pie(make_counts, labels=make_counts.index, autopct='%1.1f%%', startangle=90, colors=plt.cm.Set3.colors)
plt.title("Car Make Distribution")
plt.axis('equal')  # To make it a perfect circle
plt.show()
```

### Question 1: What are the most common car colours?

```python
color_counts = cars['Colour'].value_counts()
# Plot
plt.figure(figsize=(6, 6))
plt.pie(color_counts, labels=color_counts.index, autopct='%1.1f%%', startangle=90, colors=plt.cm.Pastel1.colors)
plt.title('Car Colour Distribution')
plt.axis('equal')
plt.show()
```

### Question 2: What is the relationship between car make and price?

```python
# Clean price
cars["Price"] = cars["Price"].replace({r'\$': '', ',': ''}, regex=True).astype(float)

avg_price = cars.groupby("Make")["Price"].mean()
avg_price.plot(kind='bar', color='skyblue')
plt.title('Average Car Price by Make')
plt.ylabel('Price (USD)')
plt.xlabel('Car Make')
plt.grid(axis='y')
plt.show()
```

### Question 3: Which car make has the highest average odometer reading?

```python
avg_odometer = cars.groupby("Make")["Odometer (KM)"].mean()

avg_odometer.sort_values().plot(kind='barh', color='lightgreen')
plt.title('Average Odometer by Car Make')
plt.xlabel('Odometer (KM)')
plt.ylabel('Car Make')
plt.grid(axis='x')
plt.show()
```
